# 07. Publication-Ready VisualizationsFour publication-quality (300 DPI) figures: correlation heatmap, violin/box distributions by tier, regression slopes with 95% CI bands, mediation flow diagram.

```python
import pandas as pd, numpy as npimport matplotlib.pyplot as pltimport seaborn as snsdf = pd.read_csv('/mnt/data/cleaned_dataset.csv')features = ['Access_Plus', 'Bank_Barrier', 'Resilience', 'Motivation', 'GPA']plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')corr = df[features].corr()mask = np.triu(np.ones_like(corr, dtype=bool))fig, ax = plt.subplots(figsize=(8, 6), dpi=300)sns.heatmap(corr, mask=mask, annot=True, fmt='.3f', cmap='vlag', center=0, ax=ax, square=True)ax.set_title('Figure 1: Pearson Correlation Matrix', fontsize=12, fontweight='bold', pad=12)plt.tight_layout()plt.savefig('/mnt/data/Fig1_Correlation_Heatmap.png', dpi=300)plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9), dpi=300)for ax, col in zip(axes.flat, ['Bank_Barrier', 'Resilience', 'Motivation', 'GPA']):    sns.violinplot(data=df, x='Access_Tier', y=col, ax=ax, inner='box', ax.setSet2')    ax.set_title(col, fontweight='bold')plt.suptitle('Figure 2: Distributions by Access Tier', fontsize=13, fontweight='bold', y=1.0)plt.tight_layout()plt.savefig('/mnt/data/Fig2_Distributions_by_Tier.png', dpi=300)plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5), dpi=300)sns.regplot(data=df, x='Motivation', y='GPA', ax=axes[0], color='#1f77b4',            scatter_kws={'alpha': 0.7, 's': 40}, line_kws={'linewidth': 2})axes[0].set_title('(A) Motivation vs GPA', fontweight='bold')sns.regplot(data=df, x='Access_Plus', y='GPA', ax=axes[1], color='#d62728',            scatter_kws={'alpha': 0.7, 's': 40}, line_kws={'linewidth': 2})axes[1].set_title('(B) Premium AI Access vs GPA (Resilience Paradox)', fontweight='bold')plt.suptitle('Figure 3: Regression Slopes with 95% CI', fontsize=13, fontweight='bold', y=1.02)plt.tight_layout()plt.savefig('/mnt/data/Fig3_Regression_Slopes.png', dpi=300)plt.show()
```
